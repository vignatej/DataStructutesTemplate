```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("housing.csv")
df = df.dropna()
```

```python
df.head(5)
```

```python
cat_labels, num_labels=[],[]
```

```python
dtype_ser = df.dtypes
dtype_inx = dtype_ser.index
for i in range(len(dtype_ser)):
    if(dtype_ser.iloc[i] in ["object"]):
        cat_labels.append(dtype_inx[i])
    else:
        num_labels.append(dtype_inx[i])
```

```python
cat_labels, num_labels
```

```python
from sklearn.preprocessing import LabelEncoder

enc_dict = dict()
for i in cat_labels:
    enc = LabelEncoder()
    enc.fit(df[i])
    # enc_dict[i] = enc
    df[i] = enc.transform(df[i])
    # print(x_train[i])
df
```

```python
from sklearn.preprocessing import StandardScaler

scalar = StandardScaler()
df[num_labels] = scalar.fit_transform(df[num_labels])
```

```python
x = df.drop("median_house_value", axis=1)
y = df[["median_house_value"]]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y)
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)

pred = model.predict(x_test)
```

```python
pred, y_test
```

```python
from sklearn.metrics import mean_absolute_error

mean_absolute_error(y_test, pred)
```

```python
df = pd.read_csv("housing.csv")
df = df.dropna()
from sklearn.model_selection import train_test_split
x, y = df.drop(["ocean_proximity"], axis=1), df["ocean_proximity"]
x_train, x_test, y_train, y_test = train_test_split(x, y)
from sklearn.ensemble import RandomForestClassifier
m = RandomForestClassifier()
m.fit(x_train, y_train)
pred = m.predict(x_test)
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pred)
```

```python
from sklearn.model_selection import GridSearchCV

params = {
    # "ccp_alpha":["l2", "l1"],
    'max_depth': [10, 50, 100]
}
grid = GridSearchCV(RandomForestClassifier(), params, cv=5)
grid.fit(x_train, y_train)

accuracy_score(y_test, grid.best_estimator_.predict(x_test))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, grid.best_estimator_.predict(x_test))
```

```python
from sklearn.svm import SVC
model = SVC()
model.fit(x_train, y_train)
accuracy_score(y_test, model.predict(x_test))
```
