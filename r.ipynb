```python
d3=pd.DataFrame(
    {"key1":["a","b","c","c","d","e"],
     "num1":range(6)})
d4=pd.DataFrame(
    {"key2":["b","c","e","f"],
     "num2":range(4)})
```

```python
mg = pd.merge(d3, d4, left_on="key1", right_on="key2", how="left")
```

```python
df=pd.DataFrame(
    {"key1":list("aabbab"),
     "key2":["one","two","three"]*2,
     "data1":np.random.randn(6),
     "data2":np.random.randn(6)})
df
```

```python
for name, grp in df.groupby("key1"):
    print(name)
```

```python
for name, grp in df.groupby("key1"):
    print(grp)
```

```python
for name, grp in df.groupby("key1"):
    print(grp["data1"].mean(), grp["data1"].max(), grp["data1"].idxmax())
```
